## Qeustion 6

```python
horsepower_frame = data['horsepower']
median_horsepower = horsepower_frame.median()
print(f"Median horsepower: {median_horsepower}")
```

```python
horsepower_frame.value_counts()
```

```python
most_freq = horsepower_frame.value_counts().index[0]
print(f"Most frequent horsepower: {most_freq}")
```

```python
horsepower_fill_frame = horsepower_frame.fillna(most_freq)
median_fill_horsepower = horsepower_fill_frame.median()
print(f"Median horsepower after filling: {median_fill_horsepower}")
```

```python
print('Median value has decreased: ' + str(median_horsepower > median_fill_horsepower))
```
